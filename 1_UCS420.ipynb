```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])

print(arr[::-1])
```

```python
array1 = np.array([[1, 2, 3],
                   [2, 4, 5],
                   [1, 2, 3]])

a = array1.flatten()
b = array1.ravel()

print(a)
print(b)
```

```python
arr1 = np.array([[1, 2],
                 [3, 4]])

arr2 = np.array([[1, 2],
                 [3, 4]])

print(arr1 == arr2)
```

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

value, count = np.unique(x, return_counts=True)

m = count.max()

for i in range(len(count)):
    if count[i] == m:
        print(value[i])
        print(np.where(x == value[i])[0])
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

value, count = np.unique(y, return_counts=True)

m = count.max()

for i in range(len(count)):
    if count[i] == m:
        print(value[i])
        print(np.where(y == value[i])[0])
```

```python
gfg = np.matrix([[4, 1, 9],
                 [12, 3, 1],
                 [4, 5, 6]])

print(np.sum(gfg))

print(np.sum(gfg, axis=1))

print(np.sum(gfg, axis=0))
```

```python
n_array = np.array([[55, 25, 15],
                    [30, 44, 2],
                    [11, 45, 77]])

print(np.trace(n_array))

value, vector = np.linalg.eig(n_array)

print(value)
print(vector)

print(np.linalg.inv(n_array))

print(np.linalg.det(n_array))
```

```python
x = np.array([[2, 3, 4],
              [3, 2, 9]])

y = np.array([[1, 5, 0],
              [5, 10, 3]])

print(np.inner(x, y))

print(np.outer(x, y))

print(np.cross(x, y))
```

```python
array = np.array([[1, -2, 3],
                  [-4, 5, -6]])

print(np.abs(array))
```

```python
print(np.percentile(array, 25))

print(np.percentile(array, 50))

print(np.percentile(array, 75))

print(np.percentile(array, 25, axis=0))

print(np.percentile(array, 50, axis=0))

print(np.percentile(array, 75, axis=0))

print(np.percentile(array, 25, axis=1))

print(np.percentile(array, 50, axis=1))

print(np.percentile(array, 75, axis=1))
```

```python
print(np.mean(array))
print(np.median(array))
print(np.std(array))

print(np.mean(array, axis=0))
print(np.median(array, axis=0))
print(np.std(array, axis=0))

print(np.mean(array, axis=1))
print(np.median(array, axis=1))
print(np.std(array, axis=1))
```

```python
a = np.array([-1.8, -1.6, -0.5, 0.5, 1.6, 1.8, 3.0])

print(np.floor(a))

print(np.ceil(a))

print(np.trunc(a))

print(np.round(a))
```
